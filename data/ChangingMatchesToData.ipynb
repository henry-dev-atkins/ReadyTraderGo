```python
import pandas as pd
import numpy as np
import csv
```

```python
def FileConversion(inCSV, outCSV, format = 'Jamie'):
	if format=='New':
		with open(inCSV, 'r') as csvfile1:
			with open(outCSV, 'w', newline='') as csvfile2:
				reader = csv.reader(csvfile1)
				writer = csv.writer(csvfile2)
				writer.writerow(['Time', 'Instrument', 'Operation', 'OrderId', 'Side', 'Volume', 'Price', 'Lifespan'])
				for row in reader:
					if row[0] == 'Time':
						continue
					if row[2] == 'Cancel':
						continue
					if row[2] == 'Amend':
						continue
					if row[2] == 'Hedge':
						continue
					#Time,Competitor,Operation,OrderId,Instrument,Side,Volume,Price,Lifespan,Fee
					if row[2] == 'Trade':
						time = float(row[0])
						instrument = int(row[4])
						operation = 'Trade'
						order_id = int(row[3])
						side = row[5]
						volume = int(row[6])
						price = int(row[7])//100
						lifespan = row[8]
						continue
					
					else:
						try:time = float(row[0])
						except:time = row[0]
						try:instrument = int(row[4])
						except:instrument = 0
						operation = row[2]
						try:order_id = int(row[3])
						except:order_id = row[3]
						side = row[5]
						try:volume = int(row[6])
						except:volume = row[6]
						try:price = int(row[7])//100
						except:price = row[7]
						lifespan = row[8]

					# Write a new row to the output CSV file
					writer.writerow([time, instrument, operation, order_id, side, volume, price, lifespan])
	if format=='Jamie':
		with open(inCSV, 'r') as csvfile1:
			with open(outCSV, 'w', newline='') as csvfile2:
				reader = csv.reader(csvfile1)
				writer = csv.writer(csvfile2)
				writer.writerow(['Time', 'Instrument', 'Operation', 'OrderId', 'Side', 'Volume', 'Price', 'Lifespan'])
				for row in reader:
					if row[0] == 'Time':
						continue
					if row[2] == 'Cancel':
						continue
					if row[2] == 'Amend':
						continue
					if row[2] == 'Hedge':
						continue
					#Time,Instrument,Operation,OrderId,Side,Volume,Price,Lifespan
					if row[2] == 'Trade':
						time = float(row[0])
						instrument = int(row[1])
						operation = 'Trade'
						order_id = int(row[3])
						side = row[4]
						volume = int(row[5])
						price = int(row[6])//100
						lifespan = row[7]
						continue
					
					else:
						try:time = float(row[0])
						except:time = row[0]
						try:instrument = int(row[1])
						except:instrument = row[1]
						operation = row[2]
						try:order_id = int(row[3])
						except:order_id = row[3]
						side = row[4]
						try:volume = int(row[5])
						except:volume = row[5]
						try:price = int(row[6])//100
						except:price = row[6]
						lifespan = row[7]

					# Write a new row to the output CSV file
					writer.writerow([time, instrument, operation, order_id, side, volume, price, lifespan])
```

```python
FileConversion("JamiesData\match23_events_marketized.csv", "market_data_Jamie_23.csv", 'Jamie')
```
